### 1. Distributions of Numerical Variables (Histograms and Box Plots)

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_cols = ['Customer_Rating', 'Preparation_Time_Min', 'Rider_Distance_KM']

plt.figure(figsize=(15, 5))
for i, col in enumerate(numerical_cols):
    plt.subplot(1, 3, i + 1)
    sns.histplot(df[col], kde=True, bins=10)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 5))
for i, col in enumerate(numerical_cols):
    plt.subplot(1, 3, i + 1)
    sns.boxplot(y=df[col], palette='pastel')
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 2. Categorical Variable Counts (Bar Plots)

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.countplot(x='City', data=df, palette='deep')
plt.title('Count of Orders by City')
plt.xlabel('City')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(1, 2, 2)
sns.countplot(x='Cuisine', data=df, palette='deep')
plt.title('Count of Orders by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 3. Relationships Between Categorical and Numerical Variables (Box Plots)

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
sns.boxplot(x='City', y='Customer_Rating', data=df, palette='viridis')
plt.title('Customer Rating by City')
plt.xlabel('City')
plt.ylabel('Customer Rating')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(2, 2, 2)
sns.boxplot(x='Cuisine', y='Customer_Rating', data=df, palette='viridis')
plt.title('Customer Rating by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Customer Rating')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(2, 2, 3)
sns.boxplot(x='City', y='Preparation_Time_Min', data=df, palette='plasma')
plt.title('Preparation Time by City')
plt.xlabel('City')
plt.ylabel('Preparation Time (Min)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(2, 2, 4)
sns.boxplot(x='Cuisine', y='Preparation_Time_Min', data=df, palette='plasma')
plt.title('Preparation Time by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Preparation Time (Min)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
sns.boxplot(x='City', y='Rider_Distance_KM', data=df, palette='magma')
plt.title('Rider Distance by City')
plt.xlabel('City')
plt.ylabel('Rider Distance (KM)')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(1, 2, 2)
sns.boxplot(x='Cuisine', y='Rider_Distance_KM', data=df, palette='magma')
plt.title('Rider Distance by Cuisine Type')
plt.xlabel('Cuisine Type')
plt.ylabel('Rider Distance (KM)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 4. Relationships Between Two Numerical Variables (Scatter Plots and Correlation Heatmap)

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(x='Avg_Meal_Price_INR', y='Customer_Rating', data=df, hue='City', palette='deep')
plt.title('Avg Meal Price vs Customer Rating')
plt.xlabel('Average Meal Price (INR)')
plt.ylabel('Customer Rating')
plt.grid(linestyle='--', alpha=0.7)

plt.subplot(1, 2, 2)
sns.scatterplot(x='Preparation_Time_Min', y='Rider_Distance_KM', data=df, hue='City', palette='deep')
plt.title('Preparation Time vs Rider Distance')
plt.xlabel('Preparation Time (Min)')
plt.ylabel('Rider Distance (KM)')
plt.grid(linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# Calculate the correlation matrix for numerical columns
correlation_matrix = df[numerical_cols + ['Avg_Meal_Price_INR']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
